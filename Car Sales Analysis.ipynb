```sql
SELECT 
    brand,
    COUNT(*) as total_cars,
    ROUND(AVG(price), 2) as avg_price
FROM hackathon_db.car_sales_clean
GROUP BY brand
ORDER BY total_cars DESC
LIMIT 10;
```

```sql
SELECT 
    CASE 
        WHEN car_age <= 2 THEN '0-2 years'
        WHEN car_age <= 5 THEN '3-5 years'
        WHEN car_age <= 10 THEN '6-10 years'
        ELSE '11+ years'
    END as age_group,
    ROUND(AVG(price), 2) as avg_price,
    COUNT(*) as cars
FROM hackathon_db.car_sales_clean
GROUP BY 
    CASE 
        WHEN car_age <= 2 THEN '0-2 years'
        WHEN car_age <= 5 THEN '3-5 years'
        WHEN car_age <= 10 THEN '6-10 years'
        ELSE '11+ years'
    END
ORDER BY age_group;
```

```sql
SELECT 
    mileage,
    price,
    brand
FROM hackathon_db.car_sales_clean
WHERE price < 100000
LIMIT 500;
```

```sql
SELECT 
    fuel_type,
    COUNT(*) as count,
    ROUND(AVG(price), 2) as avg_price
FROM hackathon_db.car_sales_clean
GROUP BY fuel_type
ORDER BY count DESC;
```

```sql
SELECT 
    price_category,
    COUNT(*) as count,
    ROUND(AVG(price), 2) as avg_price,
    ROUND(MIN(price), 2) as min_price,
    ROUND(MAX(price), 2) as max_price
FROM hackathon_db.car_sales_clean
GROUP BY price_category
ORDER BY avg_price;
```

```sql
SELECT 
    brand,
    model,
    year,
    price,
    mileage
FROM hackathon_db.car_sales_clean
ORDER BY price DESC
LIMIT 10;
```
